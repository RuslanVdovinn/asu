```python
from PIL import ImageDraw, ImageFile


class NewImageDraw:
    def __init__(self, img):
        self.img: ImageFile = img

    def triangle(self, xys: tuple, fill):
        length = int(xys[2])
        x = int((length**2 - (length / 2) ** 2) ** 0.5)
        draw = ImageDraw.Draw(self.img)
        draw.polygon(
            [
                (xys[0] - x, xys[1] + length // 2),
                (xys[0] - x, xys[1] - length // 2),
                (xys[0], xys[1]),
            ],
            fill=fill,
        )
```

```python
from PIL import Image

im = Image.open("test.jpg")
dr = NewImageDraw(im)
dr.triangle((150, 100, 150), (255, 0, 0))
# im.show()
im.save("image.bmp")
```
