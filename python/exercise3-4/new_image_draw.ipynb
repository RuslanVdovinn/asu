```python
from PIL import ImageDraw, ImageFile


class NewImageDraw:
    def __init__(self, img):
        self.img: ImageFile = img  # Исходное изображение

    def triangle(self, xys: tuple, fill):
        length = xys[2]  # Длина равностороннего треугольника
        x = (length**2 - (length / 2) ** 2) ** 0.5  # Теорема пифагора
        draw = ImageDraw.Draw(self.img)
        # По x отнимаем катет, по y смещаем на половину длинны стороны
        draw.polygon(
            [
                (xys[0] - int(x), xys[1] + length // 2),
                (xys[0] - int(x), xys[1] - length // 2),
                (xys[0], xys[1]),
            ],
            fill=fill,
        )
```

```python
from PIL import Image

im = Image.open("test.jpg")
dr = NewImageDraw(im)
dr.triangle((150, 100, 150), (255, 0, 0))
# im.show()
im.save("image.bmp")
```
