```python
class Functor:
    def __init__(self, a, b, c):  # Конструктор
        self.__a = a
        self.__b = b
        self.__c = c
        self.__hist = []

    def calc(self, x, y):  # Обычный метод
        result = (self.__a * x) / (self.__b * y) + self.__c
        self.__hist.append((x, y, result))
        return result

    def history(self):  # Возвращает историю вычислений в виде кортежа
        return tuple(self.__hist)

    def max_result(self):  # Вычисляет максимальный результат
        max = None
        for h in self.__hist:
            if max == None or max < h[-1]:
                max = h[-1]  # Берем последнее значение из кортежа
        return max

    def __iter__(self):  # Позволяет итерироваться по внутреннему списку
        return self.__hist

    def __str__(self):  # Выводит описание в нужном формате
        return f"z = ({self.__a} * x) / ({self.__b} * y) + {self.__c}"


_original_max = max  # Сохраняем функцию


def max(*args, **kwargs):
    # Проверяем что у объекта есть атрибут max_result
    if len(args) == 1 and hasattr(args[0], "max_result"):
        return args[0].max_result()
    return _original_max(*args, **kwargs)  # Иначе стандартная max
```

```python
f = Functor(2, 1, 3)
print(f.calc(3, 2))
print(f.history())
print(f)
```
