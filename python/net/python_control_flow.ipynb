# Управляющие конструкции

## 1. Операторы сравнения, условные конструкции и логические операторы

## Постановка задачи

Представим, что мы пишем маленькую программу-калькулятор.

Сначала она умеет только выполнять одну операцию.  
Потом программа научится:

- выбирать действие;
- проверять условия;
- повторять вычисления;
- использовать функции;
- принимать разные аргументы.

Так мы увидим, зачем вообще нужны управляющие конструкции.

### Линейная программа

До управляющих конструкций программа выполняется **строго сверху вниз**.

Схема:

```text
команда 1
   ↓
команда 2
   ↓
команда 3
```

Это удобно для простых вычислений, но недостаточно для реальных программ.

````python
a = 10
b = 5

result = a + b

result
````

В этом примере Python просто выполняет команды по порядку.

Но если мы захотим выбирать операцию — сложение, вычитание или умножение — линейного выполнения уже недостаточно.

## Логический тип данных

Условия в Python строятся вокруг логического типа данных — `bool`.

У него всего два значения:

- `True` — истина;
- `False` — ложь.

Любое условие в итоге должно превратиться в `True` или `False`.

````python
True
````

Проверим тип значения `True`.

````python
type(True)
````

`False` — второе возможное значение булева типа.

````python
False
````

## Операторы сравнения

Операторы сравнения возвращают `True` или `False`.

| Оператор | Значение |
|---|---|
| `==` | равно |
| `!=` | не равно |
| `>` | больше |
| `<` | меньше |
| `>=` | больше или равно |
| `<=` | меньше или равно |

Проверим простое сравнение чисел.

````python
10 > 5
````

Проверка на равенство записывается через двойной знак `==`.

````python
10 == 10
````

**Важно не путать:**

- `=` — присваивание значения переменной;
- `==` — проверка на равенство.

````python
score = 10

score == 10
````

Проверка на неравенство записывается через `!=`.

````python
10 != 5
````

## Условные конструкции

Условная конструкция позволяет программе выбрать, какой код выполнить.

**Схема:**

```text
условие?
 ├── True  → выполнить блок if
 └── False → пропустить блок if
```

### Условный оператор `if`

`if` выполняет блок кода, только если условие истинно.

````python
temperature = 25

if temperature > 20:
    print("На улице тепло")
````

### Условная конструкция `if-else`

`else` задаёт альтернативную ветку.

**Схема:**

```text
условие?
 ├── True  → выполнить if
 └── False → выполнить else
```

````python
temperature = 15

if temperature > 20:
    print("Тепло")
else:
    print("Прохладно")
````

### Условный оператор `elif`

`elif` используется, когда вариантов больше двух.

Python проверяет условия сверху вниз и выполняет первый подходящий блок.

````python
score = 82

if score >= 90:
    print("Отлично")
elif score >= 70:
    print("Хорошо")
else:
    print("Нужно подтянуть знания")
````

### Как Python читает конструкцию `if-elif-else`

```text
score >= 90 ?
 ├── True  → Отлично
 └── False ↓

score >= 70 ?
 ├── True  → Хорошо
 └── False → Нужно подтянуть знания
```

Как только Python нашёл подходящую ветку, остальные условия уже не проверяются.

**Практический пример:**
- доступ к фильму.

Представим онлайн-кинотеатр. Чтобы смотреть фильм, пользователь должен быть старше 18 лет.

````python
age = 20

if age >= 18:
    print("Фильм доступен")
else:
    print("Фильм недоступен")
````

## Логические операторы

Иногда одного условия недостаточно.

**Пример:**

- возраст пользователя должен быть не меньше 18;
- и у пользователя должен быть билет.

Для объединения условий используются логические операторы.

### Логический оператор `and`

`and` возвращает `True`, только если **оба условия истинны**.

```text
True  and True  → True
True  and False → False
False and True  → False
False and False → False
```

````python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("Вход разрешён")
else:
    print("Вход запрещён")
````

### Логический оператор `or`

`or` возвращает `True`, если **хотя бы одно условие истинно**.

````python
is_weekend = False
is_holiday = True

if is_weekend or is_holiday:
    print("Можно отдыхать")
else:
    print("Рабочий день")
````

### Логический оператор `not`

`not` меняет значение условия на противоположное.

````python
is_raining = False

if not is_raining:
    print("Можно идти гулять")
````

### Цепочка сравнений

В Python можно писать сравнения так же, как в математике.

````python
age = 25

18 <= age <= 65
````

Это читается как:

> возраст находится между 18 и 65.

Такой синтаксис часто делает код понятнее.

## Вложенные условия

Условия можно вкладывать друг в друга.

Но лучше не злоупотреблять вложенностью: слишком глубокий код сложнее читать.

````python
age = 20
has_ticket = True

if age >= 18:
    if has_ticket:
        print("Вход разрешён")
    else:
        print("Нужен билет")
else:
    print("Возрастное ограничение")
````

Тот же пример часто можно записать проще через `and`.

````python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("Вход разрешён")
else:
    print("Вход запрещён")
````

## Практика. Первая версия калькулятора

Теперь используем условия в примере. У нас есть два числа и выбранная операция.

````python
a = 10
b = 5
operation = "+"
````

Здесь видно, зачем нужны условия: программа выбирает действие.

# 2. Циклы, функции, объекты и передача аргументов

## Цикл `while`

Цикл позволяет повторять действия.

Цикл `while` используется, когда мы хотим повторять код, **пока условие истинно**.

**Схема:**

```text
проверить условие
        ↓
если True → выполнить тело цикла
        ↓
вернуться к проверке
        ↓
если False → выйти из цикла
```

````python
counter = 1

while counter <= 5:
    print(counter)
    counter += 1
````

Здесь `counter += 1` увеличивает счётчик на 1.

Без этой строки условие всегда оставалось бы истинным, и цикл стал бы бесконечным.

### Типичная ошибка — бесконечный цикл

Такой код запускать не будем, но важно понимать проблему:

```python
counter = 1

while counter <= 5:
    print(counter)
```

Значение `counter` не меняется, поэтому условие всегда истинно.

**Пример использования цикла `while`:**
- уменьшаем баланс.

Представим, что пользователь тратит деньги, пока баланс больше нуля.

````python
balance = 100
payment = 30

while balance > 0:
    print("Текущий баланс:", balance)
    balance -= payment

print("Баланс закончился")
````

## Цикл `for`

Цикл `for` используется, когда нужно перебрать последовательность значений.

````python
for number in range(5):
    print(number)
````

### Функция `range()`

`range()` создаёт последовательность чисел.

**Основные варианты:**

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

**Важно:** правая граница не включается в последовательность.

`range(5)` даёт числа от 0 до 4.

````python
list(range(5))
````

`range(1, 6)` даёт числа от 1 до 5.

````python
list(range(1, 6))
````

Третий аргумент задаёт шаг.

````python
list(range(0, 11, 2))
````

Шаг может быть **отрицательным**.

````python
list(range(10, 0, -1))
````

**Практический пример:**
- сумма чисел.

С помощью цикла можно накапливать результат.

````python
total = 0

for number in range(1, 6):
    total += number

total
````

Схема накопления:

```text
total = 0
number = 1 → total = 1
number = 2 → total = 3
number = 3 → total = 6
...
```

**Пример:**
- таблица умножения.

Циклы подходят для повторяющихся вычислений.

````python
number = 7

for multiplier in range(1, 11):
    print(number, "*", multiplier, "=", number * multiplier)
````

## Функции

Функция — это способ упаковать набор действий под одним именем.

Функции нужны, чтобы:

- не копировать один и тот же код;
- делать программу понятнее;
- переиспользовать логику;
- разбивать большую задачу на маленькие части.

### Проблема копирования кода

Если мы много раз пишем похожие строки, код становится неудобным.

````python
print("Привет, Анна")
print("Привет, Иван")
print("Привет, Мария")
````

Лучше создать функцию, которая принимает имя.

````python
def greet(name):
    print("Привет,", name)


greet("Анна")
greet("Иван")
greet("Мария")
````

### Как устроена функция

```text
def имя_функции(аргументы):
    тело функции
```

Функция не выполняется, пока мы её не вызовем.

````python
def say_hello():
    print("Привет!")


say_hello()
````

### Функция `return`

`return` возвращает результат функции наружу.

Это важно, если результат нужно использовать дальше.

````python
def add(a, b):
    return a + b


result = add(10, 5)

result
````

**Важно:** `print` и `return` — не одно и то же

- Функция `print()` просто показывает значение на экране.

- Функция `return` позволяет сохранить результат и продолжить с ним работать.

````python
def add_print(a, b):
    print(a + b)


x = add_print(2, 3)

print("x =", x)
````

Функция `print()` ничего не возвращает, поэтому в переменной `x` оказывается `None`.

````python
def add_return(a, b):
    return a + b


x = add_return(2, 3)

print("x =", x)
print("x * 10 =", x * 10)
````

## Память

В Python переменная — это имя, которое ссылается на объект.

Упрощённо:

```text
name ───► объект в памяти
```

Это важно понимать при работе с функциями и изменяемыми объектами.

## Неизменяемые объекты

Числа и строки нельзя изменить "на месте".

Когда мы меняем значение переменной, Python создаёт новое значение и перенаправляет имя.

````python
x = 10
print(id(x))

x = x + 1
print(id(x))
````

Функция `id()` показывает уникальный идентификатор объекта в памяти.

Сейчас не нужно запоминать детали, но важно понять идею: **переменная указывает на объект**.

## Передача аргументов в функцию

Когда мы передаём значение в функцию, функция получает ссылку на объект.

````python
def change_number(value):
    value = value + 1
    return value


x = 10
new_x = change_number(x)

print("x:", x)
print("new_x:", new_x)
````

Исходная переменная `x` не изменилась, потому что числа неизменяемы.

Изменяемые объекты, например списки, подробнее разберём далее.
