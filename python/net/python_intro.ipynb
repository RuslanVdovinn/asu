Jupyter Notebook состоит из отдельных ячеек:

- Code — выполняемый Python-код.
- Markdown — текст, заметки и документация.

**Важно:** Jupyter Notebook позволяет хранить код, результаты вычислений и пояснения в одном документе.

## 3. Типы данных

В Python любое значение имеет **тип данных**. Эти типы нужны для хранения и обработки разных видов данных в программе: чисел, текста и логических условий.

**Основные (встроенные) типы данных в Python:**

- **int** (integer) — целое число (число без дробной части);
- **float** (floating point number) — число с плавающей точкой (число с десятичной дробью);
- **str** (string) — строка текста (последовательность символов, заключённая в кавычки);
- **bool** (boolean) — логический тип (принимает только два значения: True — истина, False — ложь).

### Один оператор — разное поведение

**Важно:** результат арифметической операции зависит от типа данных.


```python
10 + 5
```

```python
"10" + "5"
```

Для строк оператор `+` означает объединение текста.

```python
"Python" + " " + "Notebook"
```

Для строк оператор `*` означает повторение текста.

```python
"10" * 3
```

## 5. Переменные

**Переменная** — это имя, под которым в памяти вычислительного устройства хранится значение.

Простая аналогия: переменная — это «коробка с именем», в которую можно положить значение и потом обращаться к нему по этому имени.

**Особенности переменных в Python:**

- не нужно объявлять тип — Python сам определяет тип по присвоенному значению;
- динамическая типизация — одна и та же переменная может сначала хранить число, а потом строку;
- присваивание через = (знак равенства).

Рассмотрим примеры:

```python
age = 25
age
```

```python
type(age)
```

```python
name = "Python"
name
```

```python
temperature = 23.5
temperature
```

**Важно:** в переменной можно хранить любой тип данных.

```python
age + 5
```

```python
name + " Course"
```

### Изменение переменной

```python
counter = 1
counter
```

```python
counter = 2
counter
```
