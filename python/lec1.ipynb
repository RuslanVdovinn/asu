#### Модуль tempfile

```python
import tempfile
```

```python
with tempfile.TemporaryFile() as f:
    f.write(b"0" * 100)
    f.seek(0)
    print(len(f.read()))
```

#### Модуль csv

```python
import csv
```

```python
my_data = [(1, 2, 3), (1, 3, 4)]
with open("my.csv", "w", newline="") as f:
    writer = csv.writer(f)
    for row in my_data:
        writer.writerow(row)
```

```python
with open("my.csv", "r", newline="") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```
