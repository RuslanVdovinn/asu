```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis", "blanchefort/rubert-base-cased-sentiment")
```

```python
classifier("Я люблю API и программную инженерию")
```

```python
classifier("Я не люблю API и программную инженерию")
```

```python
classifier("Я не сильно люблю API и программную инженерию")
```

```python
classifier("Не знаю люблю ли я API и программную инженерию")
```
