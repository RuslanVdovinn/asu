14. Задание
- сгенерируйте матрицу случайного размера:
 - строк от 3 до 7
 - столбцов от 2 до 12
- сгенерируйте число N - размер выборки так, чтобы оно не превышало половину количества элементов сгенерированной матрицы, но было больше 0
- сделайте случайную выборку N элементов

```python
r, c = np.random.randint(3, 8), np.random.randint(2, 13)
a = np.random.randint(10, 100, size=(r, c))
print(a)
N = np.random.randint(1, a.size // 2 + 1)
print(N)
linear_indices = np.random.choice(a.size, size=N, replace=False)
r, c = np.unravel_index(linear_indices, a.shape)
sample = a[r, c]
print(sample)
```

18. Задача на теорию вероятности. Реализовать решение методом Монте-Карло

1. Бросаем два игральных кубика: красный и синий. Определите вероятность того, что цифры на красном и синем кубиках совпадут.
2. В этом же эксперименте с кубиками нужно найти наиболее вероятную сумму цифр на кубиках.
3. Наудачу выбирается одно число от 1 до 20. Считая все числа равновозможными, определите вероятность того, что выбранное число:
* чётно;
* делится на 3;
* делится и на 2, и на 3;
* не делится ни на 2, ни на 3;
* имеет сумму цифр 9;
* имеет сумму цифр, делящуюся на 3.

```python
N = 1_000_000
red = np.random.randint(1, 7, size=N)
blue = np.random.randint(1, 7, size=N)
matching_dice = (red == blue)
prob_match = np.mean(matching_dice)
print(f"Вероятность совпадения цифр (Монте-Карло): {prob_match:.4f} (Теория: {1/6:.4f})")

sums = red + blue
values, counts = np.unique(sums, return_counts=True)
most_probable_sum = values[np.argmax(counts)]
print(f"Наиболее вероятная сумма кубиков: {most_probable_sum} (Теория: 7)")

numbers = np.random.randint(1, 21, size=N)
prob_even = np.mean(numbers % 2 == 0)
print(f"Чётное: {prob_even:.4f} (Теория: {10/20:.2f})")

prob_div3 = np.mean(numbers % 3 == 0)
print(f"Делится на 3: {prob_div3:.4f} (Теория: {6/20:.2f})")

prob_div2_and_3 = np.mean((numbers % 2 == 0) & (numbers % 3 == 0))
print(f"Делится на 2 и на 3: {prob_div2_and_3:.4f} (Теория: {3/20:.2f})")

prob_neither = np.mean((numbers % 2 != 0) & (numbers % 3 != 0))
print(f"Не делится ни на 2, ни на 3: {prob_neither:.4f} (Теория: {7/20:.2f})")

digit_sums = (numbers // 10) + (numbers % 10)

prob_sum9 = np.mean(digit_sums == 9)
print(f"Сумма цифр равна 9: {prob_sum9:.4f} (Теория: {2/20:.2f})")

prob_sum_div3 = np.mean(digit_sums % 3 == 0)
print(f"Сумма цифр делится на 3: {prob_sum_div3:.4f} (Теория: {6/20:.2f})")
```
